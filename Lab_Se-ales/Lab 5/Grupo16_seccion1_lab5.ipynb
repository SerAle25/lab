```python
import matplotlib.pyplot as plt
from scipy import ndimage
import cv2
import numpy as np
cameraman = "cameraman.ppm"
# Cargar la imagen en blanco y negro
orgImg = cv2.imread(cameraman, cv2.IMREAD_GRAYSCALE) #recibe la ruta de la imagen

# Visualizar la imagen original
plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(orgImg, cmap='gray')

plt.show()
```

**Punto 2:**
Definir los filtros

```python
# Crear matrices de los filtros a utilizar
import numpy as np

h1 = np.array([[0, 0, 0],
               [0, 1, 0],
               [0, 0, 0]])

h2 = (1/9) *  np.array([[1, 1, 1],
                        [1, 1, 1],
                        [1, 1, 1]])

h3 = (1/273) *  np.array([[1,  4,  7,  4,  1],
                          [4, 16, 26, 16, 4],
                          [7, 26, 41, 26, 7],
                          [4, 16, 26, 16, 4],
                          [1,  4,  7,  4,  1]])

h4 = np.array([[ 0, -1,  0],
               [-1,  5, -1],
               [ 0, -1,  0]])

h5 = np.array([[-1,  0,  1],
               [-2,  0,  2],
               [-1,  0,  1]])

h6 = np.array([[-1, -2, -1],
               [ 0,  0,  0],
               [ 1,  2,  1]])

h7 = np.array([[ 0,  1,  0],
               [ 1, -4,  1],
               [ 0,  1,  0]])

h8 = np.array([[-2, -1,  0],
               [-1,  1,  1],
               [ 0,  1,  2]])
```

Aplique y visualice los filtros


```python
#COPIAR ESTA CELDA PARA APLICAR TODOS LOS FILTROS
# Aplicar el filtro
filteredImg1 = cv2.filter2D(orgImg ,cv2.CV_32F, h1) #recibe como argumentos la imagen original y el filtro

# Visualizar la imagen original y la imagen filtrada en subplots
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(orgImg, cmap='gray')

plt.subplot(1, 2, 2)
plt.title('Filtro 1')
plt.imshow(filteredImg1, cmap='gray')

plt.show()
```

La forma de la matriz h1 hace que no se modifique, debido a que es la identidad

```python
# Aplicar el filtro
filteredImg2 = cv2.filter2D(orgImg ,cv2.CV_32F, h2) #recibe como argumentos la imagen original y el filtro

# Visualizar la imagen original y la imagen filtrada en subplots
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(orgImg, cmap='gray')

plt.subplot(1, 2, 2)
plt.title('Filtro 2')
plt.imshow(filteredImg2, cmap='gray')

plt.show()
```

La forma de la matriz h2 hace un promedio entre los pixeles adyacentes, lo que hace que la imagen se vea borrosa

```python
# Aplicar el filtro
filteredImg3 = cv2.filter2D(orgImg ,cv2.CV_32F, h3) #recibe como argumentos la imagen original y el filtro

# Visualizar la imagen original y la imagen filtrada en subplots
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(orgImg, cmap='gray')

plt.subplot(1, 2, 2)
plt.title('Filtro 3')
plt.imshow(filteredImg3, cmap='gray')

plt.show()
```

La matriz h3 hace algo similar a la h2, con la diferencia de que toma mas valores en cuenta para el promedio y no todos valen lo mismo, los valores de cada uno es el valor de la matriz. 

```python
# Aplicar el filtro
filteredImg4 = cv2.filter2D(orgImg ,cv2.CV_32F, h4) #recibe como argumentos la imagen original y el filtro

# Visualizar la imagen original y la imagen filtrada en subplots
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(orgImg, cmap='gray')

plt.subplot(1, 2, 2)
plt.title('Filtro 4')
plt.imshow(filteredImg4, cmap='gray')

plt.show()
```

La matriz h6 hace un realce, como el filtro de laplass pero con un valor central mas alto

```python
# Aplicar el filtro
filteredImg5 = cv2.filter2D(orgImg ,cv2.CV_32F, h5) #recibe como argumentos la imagen original y el filtro

# Visualizar la imagen original y la imagen filtrada en subplots
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(orgImg, cmap='gray')

plt.subplot(1, 2, 2)
plt.title('Filtro 5')
plt.imshow(filteredImg5, cmap='gray')

plt.show()
```

El h5 hace como si la imagen estuviera tallado en algo
